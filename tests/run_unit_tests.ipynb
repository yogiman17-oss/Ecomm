```python
import os
import sys

import pytest

# Workspace files are read-only for bytecode / cache writes during a job run.
sys.dont_write_bytecode = True

unit_tests_dir = os.path.join(os.getcwd(), "unit")
exit_code = pytest.main([unit_tests_dir, "-v", "-p", "no:cacheprovider", "--tb=short"])

assert exit_code == 0, f"Unit tests failed (pytest exit code {exit_code}). See the test output above."
print("All unit tests passed.")
```
